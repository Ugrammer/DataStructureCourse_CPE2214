Fibonacci

```python
import time
def fiboRecursive(n):
  result = 0
  if n <=  1:
    result = n
  else:
    result = fiboRecursive(n-1) + fiboRecursive(n-2)
  return result
n = 40
start_time = time.time()
print(f"fiboRecursive({n}): {fiboRecursive(n)}")
end_time = time.time()
print(f"Execution time for fiboRecursive: {end_time - start_time:.6f} seconds")
```

```python
import time
def fiboTailRecursive(n, a=0, b=1):
  if n == 0:
    result = a
  elif n == 1:
    result = b
  else:
    result = fiboTailRecursive(n - 1, b, a + b)
  return result
n = 200
start_time = time.time()
print(f"fiboTailRecursive({n}): {fiboTailRecursive(n)}")
end_time = time.time()
print(f"Execution time for fiboTailRecursive: {end_time - start_time:.6f} seconds")
```

```python
import time
def fiboLoop(n):
  a = 0
  b = 1
  if n <= 1:
    result = n;
  else:
    for i in range(2, n+1):
      result = a + b
      a = b
      b = result
  return result
n = 200
start_time = time.time()
print(f"fiboLoop({n}): {fiboLoop(n)}")
end_time = time.time()
print(f"Execution time for fiboLoop: {end_time - start_time:.6f} seconds")
```
